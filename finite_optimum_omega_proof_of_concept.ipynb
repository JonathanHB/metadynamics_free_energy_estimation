```python
energies = np.array([0,15,0,3,6,3,6])


prefactors = np.zeros((7,7))
prefactors[0,1] = 1
prefactors[0,3] = 1
prefactors[1,6] = 1
prefactors[3,4] = 1
prefactors[4,5] = 1
prefactors[2,5] = 1

prefactors += prefactors.transpose()

plt.imshow(prefactors)
```

```python
kB = 1
T = 1
kT = kB*T
lag_time = 0.01

tpm = msm_system_construction.energies_and_prefactors_to_tpm(energies, prefactors, kT, lag_time)
plt.imshow(tpm)

eigenvectors = deeptime.markov.tools.analysis.eigenvectors(tpm)
eq_probs = np.real(eigenvectors[:,0]/np.sum(eigenvectors[:,0]))
print(eq_probs)
energies_extracted = -kT*np.log(eq_probs)
print(energies_extracted)
print(energies_extracted - np.min(energies_extracted))
```

```python
#an MSM is basically equivalent to the limit of infinitely many parallel simulations on a shared MTD grid. 
# The optimal omega and the existence of an optimum could depend on the number of simulations.


#mtd params
omega = 1 #TODO: need to set parameters in a sane way
delta_T = 6
tau_mtd = 1 #what is the correlation time of an MSM? Is it 0 (or less than a single step) by virtue of markovianity?

#MTD potential
V = np.zeros(3)

#initial state
p = np.zeros(len(energies))
p[0]=1
print(p)

p_all = []


#
t_max = 30000
for t in range(t_max):
    p = np.matmul(tpm, p)
    if t%tau_mtd == 0:
        print(V)
        p_cv = np.array([p[0]+p[3], p[1]+p[4]+p[6], p[2]+p[5]])
        #p_cv = np.array([p[0]+p[4], p[1]+p[5]+p[8], p[2]+p[6]+p[9], p[3]+p[7]])
        V += omega*np.multiply(p_cv, np.exp(-V/(delta_T*kB)))
        V_bystate = np.array([V[0],V[1],V[2],V[0],V[1],V[2],V[1]])
        #V_bystate = np.array([V[0],V[1],V[2],V[3],V[0],V[1],V[2],V[3],V[1],V[2]])
        tpm = msm_system_construction.energies_and_prefactors_to_tpm(energies+V_bystate, prefactors, kT, lag_time)

    if t%50 == 0:
        plt.plot(p, color = [t/t_max, 0, 0])

    p_all.append(p)

V_bystate = np.array([V[0],V[1],V[2],V[0],V[1],V[2],V[1]])
#V_bystate = np.array([V[0],V[1],V[2],V[3],V[0],V[1],V[2],V[3],V[1],V[2]])
Z = np.sum(np.exp(-(V_bystate + energies)/kT))
mtdweights = np.exp(-(V_bystate + energies)/kT)/Z

#labeldict = {0:"Ground", 1:"LI 1", 2:"LI 2", 3:"Metastable", 4:"RC 1", 5:"RC 2", 6:"RC 3", 7:"RC 4", 8:"U 1", 9:"U 2"}
#colordict = {0:"black", 1:"green", 2:"lime", 3: "grey", 4:"yellow", 5:"orange", 6:"red", 7:"maroon", 8:"cyan", 9:"blue"}
labeldict = {0:"Ground", 1:"LI 1", 2:"Metastable", 3:"RC 1", 4:"RC 2", 5:"RC 3", 6:"U 1"}
colordict = {0:"black", 1:"green", 2: "grey", 3:"yellow", 4:"orange", 5:"red", 6:"blue"}

plt.show()
for i in range(7):
    plt.plot([p[i] for p in p_all], label=labeldict[i], color = colordict[i])
    plt.axhline(mtdweights[i], label = '_nolegend_', color = colordict[i], linestyle="dashed")
plt.legend()
```
